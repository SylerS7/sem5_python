```python
"""Introduction to NumPy
1.NumPy is a Python library used for working with arrays.
2.NumPy is a library consisting of multidimensional array objects and a collection of routines for processing the arrays.
3.In Python we have lists that serve the purpose of arrays, but they are slow to process.
4.NumPy aims to provide an array object which works faster than traditional Python lists.
5.NumPy arrays are stored at one continuous place in memory unlike lists, so processes can access and manipulate them very efficiently."""
```

```python
'''
np.asarray()
np.asarray() is used to convert an existing Python object such as a list, tuple, or another array into a NumPy array.
Syntax:
numpy.asarray(a, dtype=None)
'''
```

```python
import numpy as np
a = [10, 20, 30, 40, 50]
arr = np.asarray(a, dtype=float)
print(arr)
```

```python
'''
np.nditer()
np.nditer() is used to iterate efficiently through the elements of a NumPy array.
It is especially useful when working with multidimensional arrays.
Syntax
np.nditer(array)
'''
```

```python
#Example 1: Iterating through a 1-D Array
import numpy as np
arr = np.array([[10, 20], [30, 40]])
print(arr)
#for x in np.nditer(arr,order='C'):
   # print(x)
```

```python
#Example 2: Iterating through a 2-D Array
import numpy as np
arr = np.array([[10, 20], [30, 40]])
print(arr)
```

```python
#Example 2: Iterating through a 2-D Array using Row major and column major order
import numpy as np
arr = np.array([[10, 20], [30, 40]])
b=np.nditer(arr, order='F')
for i in b:
    print(i)
```

```python
'''
np.frombuffer()
np.frombuffer() creates a NumPy array from data stored in a buffer-like object.
A buffer can be thought of as a block of memory containing raw binary data.
It is useful when working with:

Binary files
Byte data
Network data
Memory buffers
Raw binary data
Syntax:
np.frombuffer(buffer, dtype=float, count=-1, offset=0)
'''
```

```python
import numpy as np
a = b"Hi welcome to my class"
arr = np.frombuffer(a, dtype="S1", count=6,offset=5)
print(arr)
```

```python
'''
np.fromiter()
np.fromiter() creates a NumPy array from an iterable object.
An iterable can be:
List
Tuple
Generator
range()
Set
Other objects that can be iterated
Syntax
np.fromiter(iterable, dtype)
The dtype parameter is required because NumPy needs to know what type of elements it should create.
'''
```

```python
# read the values depends on range()
import numpy as np
arr = np.fromiter(range(5), dtype=int)
print(arr)
```

```python
a=np.fromiter([10, 20, 30], dtype=int)       # List
print(a)
b=np.fromiter((10, 20, 30), dtype=float)       # Tuple
print(b)
c=np.fromiter(range(1, 4), dtype=int)        # range
print(c)
d=np.fromiter({10, 20, 30}, dtype=int)       # Set
print(d)
np.fromiter((x for x in range(3)), dtype=int) # Generator
```

```python
'''
Initialization of array using NumPy

1.Zeros()
it is used to initialize an empty array with zeros
By default data type is float
'''
```

```python
a = np.zeros(5)
print(a)
```

```python
#if you want to interger datatype
a=np.zeros(5,dtype=int)
print(a)
# for 2D array
a = np.zeros((2, 3), dtype=int)
print(a)
```

```python
#np.ones() — initialize with ones
a = np.ones(5, dtype=int)
print(a)
```

```python
'''np.full() — initialize with a specific value. If you want every element to have the same value:'''
a = np.full(5, 7)
print(a)
a = np.full((2, 3), 5)
print(a)
```

```python
'''np.empty() — create an uninitialized array.The values are not initialized to zero. You may see arbitrary values.'''
a = np.empty(5)
print(a)
```

```python
'''np.arange() — initialize with a sequence. Similar to Python's range().'''

a = np.arange(1, 6)
print(a)

# can specify a step:

a = np.arange(0, 10, 2)
print(a)
```

```python
'''np.linspace() — initialize with evenly spaced values. Create 5 evenly spaced values between 0 and 10, including both 0 and 10.'''
a = np.linspace(0, 10, 5)
print(a)
```

```python
#np.eye() — initialize an identity matrix
a = np.eye(3,dtype=int)
print(a)
```
