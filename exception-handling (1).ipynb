## 1. Handle Division by Zero using try-except

```python
try:
    a = 10
    b = 0
    result = a / b
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

## 2. Handle Invalid Input using ValueError

```python
try:
    num = int(input("Enter an integer: "))
    print("You entered:", num)
except ValueError:
    print("Error: Invalid input. Please enter an integer.")
```

## 3. Handle Multiple Exceptions using multiple except blocks

```python
try:
    num1 = int(input("Enter the numerator: "))
    num2 = int(input("Enter the denominator: "))
    print("Result:", num1 / num2)
except ValueError:
    print("Error: Please enter valid integers.")
except ZeroDivisionError:
    print("Error: Denominator cannot be zero.")
```

## 4. Use try-except-else-finally in a calculator program

```python
try:
    a = float(input("Enter first number: "))
    operator = input("Enter operator (+, -, *, /): ")
    b = float(input("Enter second number: "))

    if operator == "+":
        result = a + b
    elif operator == "-":
        result = a - b
    elif operator == "*":
        result = a * b
    elif operator == "/":
        result = a / b
    else:
        raise ValueError("Invalid operator")

except ValueError as e:
    print("Error:", e)
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
else:
    print("Result:", result)
finally:
    print("Calculator execution completed.")
```

## 5. Handle File Not Found Exception using FileNotFoundError

```python
try:
    with open("student_details.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("Error: The file was not found.")
```

## 6. Create a Custom Exception for invalid age

```python
class InvalidAgeError(Exception):
    pass

try:
    age = int(input("Enter your age: "))
    if age < 0 or age > 150:
        raise InvalidAgeError("Age must be between 0 and 150.")
    print("Valid age:", age)
except ValueError:
    print("Error: Please enter a valid integer.")
except InvalidAgeError as e:
    print("Error:", e)
```

## 7. Create a Custom Exception for insufficient bank balance

```python
class InsufficientBalanceError(Exception):
    pass

try:
    balance = float(input("Enter account balance: "))
    withdrawal = float(input("Enter withdrawal amount: "))

    if withdrawal > balance:
        raise InsufficientBalanceError("Insufficient bank balance.")

    balance -= withdrawal
    print("Withdrawal successful.")
    print("Remaining balance:", balance)
except ValueError:
    print("Error: Please enter valid amounts.")
except InsufficientBalanceError as e:
    print("Error:", e)
```

## 8. Create and Write Student Details into a text file

```python
student_name = "Rahul"
student_age = 20
student_marks = 85

with open("student_details.txt", "w") as file:
    file.write(f"Name: {student_name}\n")
    file.write(f"Age: {student_age}\n")
    file.write(f"Marks: {student_marks}\n")

print("Student details written successfully.")
```

## 9. Read Student Details from a text file

```python
try:
    with open("student_details.txt", "r") as file:
        details = file.read()
        print(details)
except FileNotFoundError:
    print("Error: student_details.txt does not exist.")
```

## 10. Append New Student Details to an existing file

```python
new_name = "Priya"
new_age = 21
new_marks = 91

with open("student_details.txt", "a") as file:
    file.write(f"Name: {new_name}\n")
    file.write(f"Age: {new_age}\n")
    file.write(f"Marks: {new_marks}\n")
    file.write("\n")

print("New student details appended successfully.")
```

## 14. Search for a Word in a text file

```python
search_word = "Rahul"

with open("student_details.txt", "r") as file:
    text = file.read()

if search_word.lower() in text.lower():
    print(f'"{search_word}" found in the file.')
else:
    print(f'"{search_word}" not found in the file.')
```

## 15. Create a NumPy Array using np.array()

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
print("NumPy array:", arr)
```

## 16. Create Arrays of Zeros and Ones using np.zeros() and np.ones()

```python
zeros_array = np.zeros(5)
ones_array = np.ones(5)

print("Array of zeros:", zeros_array)
print("Array of ones:", ones_array)
```

## 17. Create an Array Using np.arange()

```python
arr = np.arange(1, 11)
print("Array:", arr)
```

## 18. Perform Arithmetic Operations on two NumPy arrays

```python
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([1, 2, 3, 4])

print("Addition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)
print("Division:", arr1 / arr2)
```

## 19. Find Sum, Mean, Maximum, and Minimum of an array

```python
arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```

## 20. Create and Reshape a 2D NumPy Array using reshape()

```python
arr = np.arange(1, 13)
arr_2d = arr.reshape(3, 4)

print("Original 1D array:")
print(arr)

print("\nReshaped 2D array:")
print(arr_2d)
```
