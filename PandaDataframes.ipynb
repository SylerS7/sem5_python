```python
'''
Pandas – Important Features
1.High-performance data analysis tool:
-Pandas is an open-source Python library used for data manipulation, data cleaning, and data analysis.
-It is built on top of NumPy.
2.Works with large datasets:
-Pandas can efficiently work with large amounts of structured data.
-Provides operations such as filtering, sorting, grouping, aggregation, and transformation.
-However, very large datasets that do not fit into memory may require tools such as Spark, Dask, or database systems.

3.Supports different file formats:

-Pandas can read and write data in many formats:
CSV
Excel
JSON
SQL databases
HTML
Parquet
XML and others

Examples:

pd.read_csv("data.csv")
pd.read_excel("data.xlsx")
pd.read_json("data.json")

4.Flexible data representation:
-Pandas provides flexible ways to select, modify, transform, and analyze data.
-Columns can contain different data types.

5.Tabular representation:
-Pandas represents structured data in a tabular format, similar to an Excel spreadsheet or database table.
-Rows represent records.
-Columns represent attributes/features.

Example:

Name	Age	Salary
Ravi		50000
Priya	22	60000

6.Working with missing data:
-Pandas provides several functions for handling missing values such as:
isnull()
notnull()
fillna()
dropna()

Example:

df["Age"].fillna(0)

7.Indexing:
-Pandas allows us to access specific rows and columns using indexes.
-Important methods:
        loc[] → label-based indexing
        iloc[] → position-based indexing

Example:

df.loc[0, "Name"]
df.iloc[0, 1]

8.Slicing:
Slicing allows us to select a range of rows or columns.

Example:

df.iloc[0:5]

This selects rows from position 0 to 4.

9.Subsetting:
Subsetting means selecting only the required rows or columns based on a condition.

Example:
df[df["Salary"] > 50000]
This selects employees whose salary is greater than 50000.

10.Merge and Join:
Pandas can combine data from different DataFrames.
Important functions:
merge()
join()
concat()

Example:
pd.merge(df1, df2, on="StudentID")

11.Reshaping data:
Pandas allows us to change the structure or layout of data.
Important functions:
pivot()
pivot_table()
melt()
stack()
unstack()

Example:

df.pivot(index="Name", columns="Subject", values="Marks")

12.Grouping and aggregation:
Pandas can group data and perform calculations such as:
Sum
Mean
Count
Minimum
Maximum

Example:

df.groupby("Department")["Salary"].mean()

13.Sorting:
Data can be sorted based on one or more columns.
df.sort_values("Salary")

14.Data filtering:
Pandas makes it easy to filter data based on conditions.
df[df["Age"] > 20]

15.Data transformation:
Existing data can be modified or new columns can be created.
df["AnnualSalary"] = df["Salary"] * 12
'''
```

```python
'''
2. DataFrame
A DataFrame is a two-dimensional data structure provided by the Pandas library. You can think of a DataFrame like an Excel table:

Rows → records
Columns → attributes/features
Each column can contain a different data type
Every row has an index
| Index | Name     | Age | Place | Salary |
| ----: | -------- | --: | ----- | -----: |
|     0 | Anshitha |  21 | BZA   |  70000 |
|     1 | Sananvi  |  22 | BVRM  |  50000 |
|     2 | Rishi    |  23 | VZM   | 6000000 |

Name, Age, Place, Salary → columns
0, 1, 2 → index
Each horizontal record → row
'''
```

```python
'''Import Pandas
First, import Pandas:
import pandas as pd'''
```

```python
'''Method 1: Using a Dictionary: This is one of the most common methods.'''

#Example1
import pandas as pd
data = {
    "Name": ["Anshitha", "Sananvi", "Ravi"],
    "Age": [21, 22, 23],
    "Place": ["BZA", "BVRM", "VZM"],
    "Salary": [70000, 50000, 6000000]
}
df = pd.DataFrame(data)
print(df)
df["Phone no"]= [1,2,3]
print(df)
df.loc[3] = ["Swaroopa", 25, "VZM", 20000,345]
print(df)
# or u can define like this also
'''df = pd.DataFrame({
    "Name": ["Rishi", "Anshith"],
    "Age": [21, 22]
})
print(df)'''
```

```python
#Example 2
import pandas as pd
df= pd.DataFrame({
    "StudentName": ["Rishi", "Anshi"],
    "StudentCourse" : ["Java","Python"],
    "Department" : ["BCA","MCA"]
})
print(df)
print(df.index)
print(df.columns)
print(df.shape)
```

```python
'''
Selecting Data: Data manipulation often involves selecting the required rows and columns.
'''
#Selecting one column
print(df["StudentName"])
```

```python
#Selecting multiple columns
print(df[["StudentName", "StudentCourse"]])
```

```python
#Selecting a row using loc
print(df.loc[0]) #Give me the row whose index label is 0

#Selecting a row using iloc
print(df.iloc[1]) #Give me the row at position 1
```

```python
#Example 3
data1=({
    "EmpName" : ["Hemalatha","Swathi"],
    "EmpDept" :["BCA","CSE-1"],
    "EmpNO" : [7282,8912]
})
df = pd.DataFrame(data1)
print(df)
result = df[df["EmpDept"] == "BCA"]
print(result)
```

```python
#Add new column into the existing data
df["Place"] = ["BZA", "BZA"]
print(df)

#Add new record in the existing data
df.loc[2] = ["Poornima", "BCA", 8514, "VZM"]
df.loc[3] = ["Swaroopa", "MCA", 8990, "Guntur"]
df.loc[4] = ["Anjum", "BCA", 9370, "Guntur"]

print(df)
```

```python
'''
3. Filtering Data: Filtering means selecting only the records that satisfy a condition.
'''

result = df[df["EmpDept"] == "BCA"]
print(result)
```

```python
#Select multiple columns
print(df[["EmpName", "EmpDept"]])
```

```python
#Filtering using numeric values

#Find employees whose employee number is greater than 8500:

print(df[df["EmpNO"] > 9000])
```

```python
#Multiple conditions : Suppose we want employees from BCA AND EmpNO > 8000.

result = df[(df["EmpDept"] == "BCA") & (df["EmpNO"] > 8000)]

print(result)
```

```python
# Adding a new column: Suppose we want to add employee salary:

df["Salary"] = [30000, 35000, 80000, 45000, 38000]

print(df)
```

```python
#Updating a value : Suppose Swathi changes from CSE-1 to CSE.

df.loc[1, "EmpDept"] = "CSE"

print(df)
```

```python
#Add one more column

df["Experience"]= [15,15,13,12,5]
print(df)

#Deleting a column :To remove the Salary column:axis=1 → column, axis=0 → row

#df.drop("Place", axis=1, inplace=True)
print(df.columns)



```

```python
#Delete Experience column
df.drop("Experience", axis=1, inplace=True)
print(df)
```

```python
df["Salary"] = [30000, 35000, 70000, 45000, 38000]

print(df)
```

```python
#Sorting Data : Sort by employee number

df.sort_values("EmpNO", inplace=True)

print(df)
```

```python
#Basic statistical operations : For employee numbers:

print("Maximum:", df["EmpNO"].max())
print("Minimum:", df["EmpNO"].min())
print("Average:", df["EmpNO"].mean())
print("Total:", df["EmpNO"].sum())
```

```python
#Grouping Data : groupby() is very important in Pandas. For example, count employees in each department:

print(df.groupby("EmpDept")["EmpName"].count())
```

```python
#Data visualization means representing data graphically so that patterns, comparisons, and trends can be understood easily.
#1.The most commonly used Python library is Matplotlib. Install it if necessary:

#pip install matplotlib

#Import it:

import matplotlib.pyplot as plt
```

```python
#Bar Chart — Employees by Department : First calculate the number of employees in each department:

import matplotlib.pyplot as plt
dept_count = df["EmpDept"].value_counts()

print(dept_count)
```

```python
#Createing a Bar chart.  for that we need one function called bar(x-axis, y-axis)
#X-axis values.:dept_count.index, Y-axis values :dept_count.values

plt.bar(dept_count.index, dept_count.values)

plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Employees by Department")

plt.show()
```

```python
#Pie Chart — Department Distribution : A pie chart is useful when showing proportions.

dept_count = df["EmpDept"].value_counts()

plt.pie(
    dept_count.values,
    labels=dept_count.index,
    autopct="%1.1f%%"
)

plt.title("Department Distribution")

plt.show()
```
